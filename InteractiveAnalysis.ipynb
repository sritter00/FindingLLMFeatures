## Circular PCA Analysis

Test the circular PCA visualization on days of the week or months.

````python
print("Loading GPT-2 Small...")
model = HookedTransformer.from_pretrained("gpt2-small")
print("Model loaded!")
````

````python
# Test on months
months = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
]

activations, pca_model, labels = visualize_circular_concept(
    model,
    months,
    prefix="The current month is",
    target_layer=6
)
````

````python
# Test on days of the week
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

activations, pca_model, labels = visualize_circular_concept(
    model,
    days,
    prefix="Today is",
    target_layer=6
)
````

## Manifold Discovery

Search for 2D manifolds in activation space (smaller scale for interactive use).

````python
# Extract activations (smaller sample for interactive work)
X, tokens_list = extract_activations(
    model,
    target_token_count=50_000,  # Smaller for faster iteration
    target_layer=6
)
````

````python
# Find manifolds
candidates = find_manifolds(
    X,
    tokens_list,
    num_clusters=300,  # Fewer for faster iteration
    variance_threshold=0.35,
    drop_off_threshold=1.5
)

print(f"\nFound {len(candidates)} manifold candidates")
````

````python
# Review top candidates
print_manifold_results(candidates, top_n=10)
````
